```python
import matplotlib.pyplot as plt
import pygimli as pg
import pygimli.physics.traveltime as tt
```

```python
data = tt.load("SpecFEM_Refraction_arrivals_gimli.txt", verbose=True)
print(data)
```

```python
fig, ax = plt.subplots()
lines = tt.drawFirstPicks(ax, data)
```

```python
mgr = tt.TravelTimeManager(data)

# Alternatively, one can plot a matrix plot of apparent velocities which is the
# more general function also making sense for crosshole data.

ax, cbar = mgr.showData()
```

```python
mgr.invert(secNodes=3, paraMaxCellSize=10.0, zWeight=0.2, vTop=200, vBottom=4000, verbose=1)
```

```python
mgr.showFit(firstPicks=True)
```

```python
ax, cbar = mgr.showResult(logScale=False,  cMap="plasma_r")#, #coverage=mgr.standardizedCoverage())

rays = mgr.drawRayPaths(ax=ax, color="k", lw=0.3, alpha=0.9)
```
